```python
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
from pandas import Series,DataFrame
import seaborn as sns
import palettable
from sklearn import datasets 
plt.rc('font',family='Times New Roman')
```

```python
colors = [
    None,
    palettable.tableau.BlueRed_6.mpl_colors,
    palettable.cartocolors.qualitative.Bold_9.mpl_colors[2:],
    palettable.tableau.TrafficLight_9.mpl_colors
]
```

```python
dataset = "BindingDB"
setting = "S1"
data_df = pd.read_excel("./../Source_Data/{}.xlsx".format(dataset), sheet_name="{}".format(setting))
plt.figure(dpi=400)
fig = sns.boxplot(
    data = data_df,
    x="model",
    y="value",
    hue ="metric",
    orient='v',
    width=0.5,
    dodge = "auto",
    fliersize=2,
#             notch=True,
    palette=None,
    showmeans=True,#箱图显示均值，
    meanprops = {"markersize":"2.5"},
#             saturation=0.3,#设置颜色饱和度
           )
plt.legend(loc='lower right')
fig.set_xticklabels(labels = ['RF','DNN','DeepConv-DTI','GraphDTA','MolTrans','HyperAttentionDTI','DrugBAN','AI-Bind','ColdstartCPI'],rotation = 30,fontsize = 8) # 放大横轴坐标并逆时针旋转45°
plt.savefig('./raw_figures/{}_{}.jpg'.format(dataset,setting), dpi=400,bbox_inches = 'tight')
plt.show()
```

```python
import numpy as np

mean =20.4# 均值
sig = 5

# 生成数据
data = np.random.normal(loc=mean, scale=sig, size=10)

# 均值和方差
new_mean, new_var = np.mean(data), np.sqrt(np.var(data))
print("生成的数据的均值：", new_mean)
print("生成的数据的中位数：",new_var)
# 线性变换调整均值
new_data = (data+(mean-new_mean))/100

# 打印结果
print("生成的数据的均值：", np.mean(new_data))
print("生成的数据的中位数：", np.sqrt(np.var(new_data)))
new_data = pd.DataFrame(new_data,index=None)
new_data.to_csv("./test.csv")
```
